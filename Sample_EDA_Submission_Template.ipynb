## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Data Wrangling

# Create copies of original datasets
titles_clean = titles.copy()
credits_clean = credits.copy()

# Remove duplicate rows
titles_clean.drop_duplicates(inplace=True)
credits_clean.drop_duplicates(inplace=True)

# Remove extra spaces from column names
titles_clean.columns = titles_clean.columns.str.strip()
credits_clean.columns = credits_clean.columns.str.strip()

# Convert release_year to numeric
titles_clean['release_year'] = pd.to_numeric(
    titles_clean['release_year'], errors='coerce'
)

# Convert runtime to numeric
titles_clean['runtime'] = pd.to_numeric(
    titles_clean['runtime'], errors='coerce'
)

# Handle missing values in selected columns
titles_clean['title'] = titles_clean['title'].fillna('Unknown')
titles_clean['description'] = titles_clean['description'].fillna('Not Available')

# Display cleaned dataset information
print("Cleaned Titles Shape:", titles_clean.shape)
print("Cleaned Credits Shape:", credits_clean.shape)

print("\nRemaining Missing Values in Titles:")
display(titles_clean.isnull().sum())
````

### What all manipulations have you done and insights you found?

Answer Here : The following data cleaning and manipulation steps were performed during the data wrangling process:

1. Created separate copies of the original titles and credits datasets to preserve the original data.
2. Checked and removed duplicate records from both datasets.
3. Removed unnecessary spaces from column names.
4. Converted the release_year and runtime columns into numeric data types.
5. Handled missing values in the title and description columns by replacing them with suitable values.
6. Checked the dataset shape and remaining missing values after cleaning.
