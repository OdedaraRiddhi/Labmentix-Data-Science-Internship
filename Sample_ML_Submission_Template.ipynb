## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Data Wrangling

# Create copies of original datasets
titles_clean = titles.copy()
credits_clean = credits.copy()

# Remove duplicate rows
titles_clean.drop_duplicates(inplace=True)
credits_clean.drop_duplicates(inplace=True)

# Remove extra spaces from text columns
for col in titles_clean.select_dtypes(include='object').columns:
    titles_clean[col] = titles_clean[col].str.strip()

for col in credits_clean.select_dtypes(include='object').columns:
    credits_clean[col] = credits_clean[col].str.strip()

# Convert IMDb score into numeric format
titles_clean['imdb_score'] = pd.to_numeric(
    titles_clean['imdb_score'], errors='coerce'
)

# Display cleaned dataset information
print("Titles Dataset Shape:", titles_clean.shape)
print("Credits Dataset Shape:", credits_clean.shape)

print("\nMissing Values in Titles:")
display(titles_clean.isnull().sum())

print("\nMissing Values in Credits:")
display(credits_clean.isnull().sum())

print("\nData Wrangling Completed Successfully!")
````

### What all manipulations have you done and insights you found?

Answer Here : During the data wrangling process, I performed the following operations:

1. Created copies of the original titles and credits datasets.
2. Removed duplicate records from both datasets.
3. Removed extra spaces from text columns.
4. Converted the IMDb score column into numeric format.
5. Checked missing values after cleaning.

These operations helped improve data consistency and prepare the datasets for further analysis and machine learning.

The missing values and dataset statistics were examined to understand the data quality and identify areas that require further preprocessing.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here : ### Three Hypothetical Statements

**Hypothesis 1: IMDb Score Comparison**

Movies and TV Shows have different average IMDb scores.

- Null Hypothesis (H0): There is no significant difference between the average IMDb scores of Movies and TV Shows.
- Alternative Hypothesis (H1): There is a significant difference between the average IMDb scores of Movies and TV Shows.

**Hypothesis 2: IMDb Votes and IMDb Scores**

There is a significant relationship between IMDb votes and IMDb scores.

- Null Hypothesis (H0): There is no significant correlation between IMDb votes and IMDb scores.
- Alternative Hypothesis (H1): There is a significant correlation between IMDb votes and IMDb scores.

**Hypothesis 3: Runtime Comparison**

Movies and TV Shows have different average runtimes.

- Null Hypothesis (H0): There is no significant difference between the average runtimes of Movies and TV Shows.
- Alternative Hypothesis (H1): There is a significant difference between the average runtimes of Movies and TV Shows.

These three hypotheses will be tested using statistical methods such as the Independent Samples t-test and Pearson Correlation Test, with a significance level of 0.05.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Answer Here : **Null Hypothesis (H0):**

There is no significant difference between the average IMDb scores of Movies and TV Shows on Amazon Prime.

**Alternative Hypothesis (H1):**

There is a significant difference between the average IMDb scores of Movies and TV Shows on Amazon Prime.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Perform Independent Samples T-Test

from scipy.stats import ttest_ind

# Separate IMDb scores for Movies and TV Shows
movie_scores = titles_clean[
    (titles_clean['type'] == 'MOVIE')
]['imdb_score'].dropna()

show_scores = titles_clean[
    (titles_clean['type'] == 'SHOW')
]['imdb_score'].dropna()

# Perform Welch's T-Test
t_stat, p_value = ttest_ind(
    movie_scores,
    show_scores,
    equal_var=False
)

print("Movie Average IMDb Score:", movie_scores.mean())
print("TV Show Average IMDb Score:", show_scores.mean())
print("T-Statistic:", t_stat)
print("P-Value:", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant difference between the average IMDb scores.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is no significant difference between the average IMDb scores.")
````

##### Which statistical test have you done to obtain P-Value?

Answer Here : I performed an Independent Samples T-Test (Welch's T-Test) to obtain the P-Value and compare the average IMDb scores of Movies and TV Shows.

##### Why did you choose the specific statistical test?

Answer Here : I chose the Independent Samples T-Test because I wanted to compare the average IMDb scores of two independent groups: Movies and TV Shows. Welch's T-Test is suitable because it does not assume equal variances between the two groups. It helps determine whether the difference between their average IMDb scores is statistically significant.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Answer Here : **Null Hypothesis (H0):**

There is no significant correlation between IMDb votes and IMDb scores of Amazon Prime Movies and TV Shows.

**Alternative Hypothesis (H1):**

There is a significant correlation between IMDb votes and IMDb scores of Amazon Prime Movies and TV Shows.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Perform Pearson Correlation Test

from scipy.stats import pearsonr

# Select required columns and remove missing values
correlation_data = titles_clean[
    ['imdb_votes', 'imdb_score']
].dropna()

# Perform Pearson Correlation Test
correlation, p_value = pearsonr(
    correlation_data['imdb_votes'],
    correlation_data['imdb_score']
)

print("Pearson Correlation:", correlation)
print("P-Value:", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant correlation between IMDb votes and IMDb scores.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is no significant correlation between IMDb votes and IMDb scores.")
````

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation
# Handling Missing Values & Missing Value Imputation

# Check missing values
print("Missing Values Before Imputation:")
display(titles_clean.isnull().sum())

# Visualize missing values
plt.figure(figsize=(12, 6))
sns.heatmap(titles_clean.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()

# Handle numerical missing values
for col in titles_clean.select_dtypes(include=np.number).columns:
    titles_clean[col] = titles_clean[col].fillna(
        titles_clean[col].median()
    )

# Handle categorical missing values
for col in titles_clean.select_dtypes(include='object').columns:
    titles_clean[col] = titles_clean[col].fillna('Unknown')

# Check missing values after imputation
print("Missing Values After Imputation:")
display(titles_clean.isnull().sum())
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

Answer Here : I used two missing value imputation techniques:

1. **Median Imputation:** I used the median to fill missing values in numerical columns because it is less affected by extreme values and outliers compared to the mean.

2. **Unknown Imputation:** I replaced missing values in categorical columns with 'Unknown' because it preserves the available records without removing rows and clearly identifies missing information.

These techniques help maintain the dataset structure and prepare the data for further analysis and machine learning.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments
# Handling Outliers & Outlier Treatments

# Select numerical columns
numeric_cols = titles_clean.select_dtypes(include=np.number).columns

# Visualize outliers before treatment
plt.figure(figsize=(14, 6))
sns.boxplot(data=titles_clean[numeric_cols])
plt.title('Outliers Before Treatment')
plt.xticks(rotation=45)
plt.show()

# Apply IQR method
for col in numeric_cols:
    Q1 = titles_clean[col].quantile(0.25)
    Q3 = titles_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    titles_clean[col] = titles_clean[col].clip(
        lower=lower_limit,
        upper=upper_limit
    )

# Visualize outliers after treatment
plt.figure(figsize=(14, 6))
sns.boxplot(data=titles_clean[numeric_cols])
plt.title('Outliers After Treatment')
plt.xticks(rotation=45)
plt.show()

print("Outlier treatment completed successfully!")
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Answer Here : I used the **Interquartile Range (IQR) method with the Capping (Clipping) technique** to handle outliers.

First, I calculated Q1, Q3, and IQR for numerical columns. Then, values below the lower limit and above the upper limit were capped at their respective limits using the clipping method.

I chose this technique because it reduces the impact of extreme values without deleting important data records. Box plots were also used to compare outliers before and after treatment.
