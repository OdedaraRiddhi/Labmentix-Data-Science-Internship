## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Data Wrangling

# Create copies of original datasets
titles_clean = titles.copy()
credits_clean = credits.copy()

# Remove duplicate rows
titles_clean.drop_duplicates(inplace=True)
credits_clean.drop_duplicates(inplace=True)

# Remove extra spaces from text columns
for col in titles_clean.select_dtypes(include='object').columns:
    titles_clean[col] = titles_clean[col].str.strip()

for col in credits_clean.select_dtypes(include='object').columns:
    credits_clean[col] = credits_clean[col].str.strip()

# Convert IMDb score into numeric format
titles_clean['imdb_score'] = pd.to_numeric(
    titles_clean['imdb_score'], errors='coerce'
)

# Display cleaned dataset information
print("Titles Dataset Shape:", titles_clean.shape)
print("Credits Dataset Shape:", credits_clean.shape)

print("\nMissing Values in Titles:")
display(titles_clean.isnull().sum())

print("\nMissing Values in Credits:")
display(credits_clean.isnull().sum())

print("\nData Wrangling Completed Successfully!")
````

### What all manipulations have you done and insights you found?

Answer Here : During the data wrangling process, I performed the following operations:

1. Created copies of the original titles and credits datasets.
2. Removed duplicate records from both datasets.
3. Removed extra spaces from text columns.
4. Converted the IMDb score column into numeric format.
5. Checked missing values after cleaning.

These operations helped improve data consistency and prepare the datasets for further analysis and machine learning.

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here : ### Three Hypothetical Statements

Movies and TV Shows have different average runtimes.

- Alternative Hypothesis (H1): There is a significant difference between the average runtimes of Movies and TV Shows.

These three hypotheses will be tested using statistical methods such as the Independent Samples t-test and Pearson Correlation Test, with a significance level of 0.05.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Answer Here : **Null Hypothesis (H0):**

There is no significant difference between the average IMDb scores of Movies and TV Shows on Amazon Prime.

**Alternative Hypothesis (H1):**

There is a significant difference between the average IMDb scores of Movies and TV Shows on Amazon Prime.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
# Perform Independent Samples T-Test

from scipy.stats import ttest_ind

# Separate IMDb scores for Movies and TV Shows
movie_scores = titles_clean[
    (titles_clean['type'] == 'MOVIE')
]['imdb_score'].dropna()

show_scores = titles_clean[
    (titles_clean['type'] == 'SHOW')
]['imdb_score'].dropna()

# Perform Welch's T-Test
t_stat, p_value = ttest_ind(
    movie_scores,
    show_scores,
    equal_var=False
)

print("Movie Average IMDb Score:", movie_scores.mean())
print("TV Show Average IMDb Score:", show_scores.mean())
print("T-Statistic:", t_stat)
print("P-Value:", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant difference between the average IMDb scores.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is no significant difference between the average IMDb scores.")
````

### 8. Data Splitting

````python
# Split your data to train and test. Choose Splitting ratio wisely.

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Separate input features and target variable
X = titles_selected.drop(columns=['imdb_score'])
y = titles_selected['imdb_score']

# Identify numerical and categorical columns
numerical_columns = X.select_dtypes(include=['number']).columns.tolist()
categorical_columns = X.select_dtypes(include=['object']).columns.tolist()

# Split data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Prepare numerical data
numerical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Prepare categorical data
categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine preprocessing steps
preprocessor = ColumnTransformer([
    ('num', numerical_pipeline, numerical_columns),
    ('cat', categorical_pipeline, categorical_columns)
])

# Fit preprocessing only on training data
X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

print("Data Splitting Completed Successfully!")
print("Training Data Shape:", X_train_transformed.shape)
print("Testing Data Shape:", X_test_transformed.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
````
