````python
# Chart - 3 visualization code
# Chart 3: Review Length Distribution by Customer Rating

# Convert ratings into numeric values
reviews_clean['Rating_Numeric'] = pd.to_numeric(
    reviews_clean['Rating'],
    errors='coerce'
)

# Calculate review length in words
reviews_clean['Review_Word_Count'] = (
    reviews_clean['Review']
    .fillna('')
    .astype(str)
    .str.split()
    .str.len()
)

# Remove rows without valid numeric ratings
chart_data = reviews_clean.dropna(
    subset=['Rating_Numeric']
).copy()

# Arrange ratings in ascending order
rating_order = sorted(
    chart_data['Rating_Numeric'].unique()
)

# Prepare review lengths for each rating
review_groups = [
    chart_data.loc[
        chart_data['Rating_Numeric'] == rating,
        'Review_Word_Count'
    ].values
    for rating in rating_order
]

# Create Box Plot
plt.figure(figsize=(11, 6))

plt.boxplot(
    review_groups,
    labels=[str(rating) for rating in rating_order],
    showfliers=False
)

plt.title(
    'Review Length Distribution by Customer Rating',
    fontsize=14,
    fontweight='bold'
)
plt.xlabel('Customer Rating')
plt.ylabel('Review Length (Word Count)')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Display statistical summary
review_summary = chart_data.groupby(
    'Rating_Numeric'
)['Review_Word_Count'].agg(
    ['count', 'mean', 'median']
).round(2)

print("Review Length Summary by Rating:")
display(review_summary)
````

#### Chart - 4

````python
# Chart - 4: Top 10 Restaurants by Number of Reviews

# Chart 4: Review Activity Over Time

import pandas as pd
import matplotlib.pyplot as plt

try:
    # Check whether the required column exists
    if 'Time' not in reviews_clean.columns:
        raise KeyError("The 'Time' column is not available.")

    # Convert review timestamps into datetime format
    time_data = reviews_clean['Time'].astype('string').str.strip()
    review_dates = pd.to_datetime(
        time_data,
        format='%m/%d/%Y %H:%M',
        errors='coerce'
    )

    # Keep only rows with valid dates
    valid_dates = review_dates.dropna()

    if valid_dates.empty:
        raise ValueError("No valid review dates were found.")

    # Count reviews month by month
    monthly_reviews = (
        valid_dates.dt.to_period('M')
        .value_counts()
        .sort_index()
    )

    # Include months with zero reviews
    monthly_reviews.index = monthly_reviews.index.to_timestamp()
    monthly_reviews = monthly_reviews.asfreq('MS', fill_value=0)

    # Plot monthly review activity
    plt.figure(figsize=(14, 6))
    plt.plot(
        monthly_reviews.index,
        monthly_reviews.values,
        marker='o',
        linewidth=2
    )

    plt.title('Monthly Review Activity Over Time',
              fontsize=14, fontweight='bold')
    plt.xlabel('Month')
    plt.ylabel('Number of Reviews')
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Display important summary information
    print("Valid dates:", len(valid_dates))
    print("Invalid or missing dates:",
          reviews_clean['Time'].notna().sum() - len(valid_dates))

    print("\nMonth with the highest reviews:")
    print(monthly_reviews.idxmax().strftime('%B %Y'),
          "-", monthly_reviews.max(), "reviews")

    print("\nMonth with the lowest reviews:")
    print(monthly_reviews.idxmin().strftime('%B %Y'),
          "-", monthly_reviews.min(), "reviews")

except (KeyError, ValueError) as error:
    print("Data validation error:", error)

except Exception as error:
    print("An unexpected error occurred:", error)
````

````python
# Chart - 5 visualization code

# Chart 5: Distribution of Restaurant Cost for Two

import pandas as pd
import matplotlib.pyplot as plt

try:
    # Check whether the required column exists
    if 'Cost' not in metadata_clean.columns:
        raise KeyError("The 'Cost' column is not available.")

    # Convert cost values into numeric format
    cost_data = pd.to_numeric(
        metadata_clean['Cost'].astype(str)
        .str.replace(',', '', regex=False)
        .str.strip(),
        errors='coerce'
    ).dropna()

    if cost_data.empty:
        raise ValueError("No valid cost values were found.")

    # Create a histogram to show cost distribution
    plt.figure(figsize=(11, 6))
    plt.hist(cost_data, bins=10, edgecolor='black')

    plt.title('Distribution of Restaurant Cost for Two',
              fontsize=14, fontweight='bold')
    plt.xlabel('Approximate Cost for Two (₹)')
    plt.ylabel('Number of Restaurants')
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Display descriptive statistics
    print("Valid cost records:", len(cost_data))
    print("Average cost for two: ₹", round(cost_data.mean(), 2))
    print("Minimum cost: ₹", cost_data.min())
    print("Maximum cost: ₹", cost_data.max())
    print("Most common cost: ₹", cost_data.mode().iloc[0])

except (KeyError, ValueError) as error:
    print("Data validation error:", error)

except Exception as error:
    print("An unexpected error occurred:", error)
````

````python
# Chart - 6 visualization code

# Chart 6: Top 10 Most Popular Cuisines

import pandas as pd
import matplotlib.pyplot as plt

try:
    # Check whether the required column exists
    if 'Cuisines' not in metadata_clean.columns:
        raise KeyError("The 'Cuisines' column is not available.")

    # Split multiple cuisines and count each cuisine separately
    cuisine_data = (
        metadata_clean['Cuisines']
        .dropna()
        .astype(str)
        .str.split(',')
        .explode()
        .str.strip()
    )

    # Remove empty cuisine values
    cuisine_data = cuisine_data[cuisine_data != '']

    if cuisine_data.empty:
        raise ValueError("No valid cuisine values were found.")

    # Count the number of restaurants offering each cuisine
    cuisine_counts = cuisine_data.value_counts()

    # Select the top 10 cuisines
    top_10_cuisines = cuisine_counts.head(10).sort_values()

    # Create a horizontal bar chart
    plt.figure(figsize=(11, 6))
    plt.barh(
        top_10_cuisines.index,
        top_10_cuisines.values,
        edgecolor='black'
    )

    plt.title('Top 10 Most Popular Cuisines',
              fontsize=14, fontweight='bold')
    plt.xlabel('Number of Restaurants')
    plt.ylabel('Cuisine')
    plt.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Display the top 10 cuisine counts
    print("Top 10 Most Popular Cuisines:")
    display(top_10_cuisines.sort_values(ascending=False).to_frame(
        name='Restaurant Count'
    ))

    print("Total unique cuisines:", cuisine_data.nunique())
    print("Total cuisine entries:", len(cuisine_data))

except (KeyError, ValueError) as error:
    print("Data validation error:", error)

except Exception as error:
    print("An unexpected error occurred:", error)
````

````python
# Chart - 9 visualization code

# Chart 9: Relationship Between Restaurant Cost and Average Rating

import pandas as pd
import matplotlib.pyplot as plt

try:
    # Check required columns
    if 'Restaurant' not in reviews_clean.columns:
        raise KeyError("Restaurant column is missing in reviews_clean.")

    if 'Rating' not in reviews_clean.columns:
        raise KeyError("Rating column is missing in reviews_clean.")

    if 'Name' not in metadata_clean.columns or 'Cost' not in metadata_clean.columns:
        raise KeyError("Name or Cost column is missing in metadata_clean.")

    # Prepare review data
    review_data = reviews_clean.copy()
    review_data['Restaurant_Key'] = (
        review_data['Restaurant'].astype(str).str.strip()
    )

    review_data['Rating_Numeric'] = pd.to_numeric(
        review_data['Rating'], errors='coerce'
    )

    # Calculate average rating for each restaurant
    average_ratings = (
        review_data.groupby('Restaurant_Key')['Rating_Numeric']
        .agg(Average_Rating='mean', Review_Count='count')
        .reset_index()
    )

    # Prepare restaurant metadata
    restaurant_data = metadata_clean.copy()
    restaurant_data['Restaurant_Key'] = (
        restaurant_data['Name'].astype(str).str.strip()
    )

    restaurant_data['Cost_Numeric'] = pd.to_numeric(
        restaurant_data['Cost'].astype(str)
        .str.replace(',', '', regex=False)
        .str.strip(),
        errors='coerce'
    )

    # Merge both datasets using restaurant names
    merged_data = pd.merge(
        restaurant_data,
        average_ratings,
        on='Restaurant_Key',
        how='inner'
    )

    # Keep valid cost and rating values
    chart_data = merged_data.dropna(
        subset=['Cost_Numeric', 'Average_Rating']
    ).copy()

    if chart_data.empty:
        raise ValueError("No matching valid restaurant data found.")

    # Calculate correlation
    correlation = chart_data['Cost_Numeric'].corr(
        chart_data['Average_Rating']
    )

    # Create scatter plot
    plt.figure(figsize=(10, 6))
    plt.scatter(
        chart_data['Cost_Numeric'],
        chart_data['Average_Rating'],
        alpha=0.7
    )

    plt.title('Restaurant Cost vs Average Customer Rating',
              fontsize=14, fontweight='bold')
    plt.xlabel('Approximate Cost for Two (₹)')
    plt.ylabel('Average Customer Rating')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Display summary
    print("Number of matched restaurants:", len(chart_data))
    print("Average restaurant cost: ₹",
          round(chart_data['Cost_Numeric'].mean(), 2))
    print("Average customer rating:",
          round(chart_data['Average_Rating'].mean(), 2))
    print("Correlation between cost and rating:",
          round(correlation, 2))

except (KeyError, ValueError) as error:
    print("Data validation error:", error)

except Exception as error:
    print("An unexpected error occurred:", error)
````

###Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

**Answer Here :** Three Hypothetical Statements

**Hypothesis 1: Customer Ratings and Pictures**

**Hypothesis 2: Review Length and Pictures**

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Answer Here :** **Null Hypothesis (H₀):**

There is no significant difference in customer ratings between reviews with pictures and reviews without pictures.

**Alternative Hypothesis (H₁):**

There is a significant difference in customer ratings between reviews with pictures and reviews without pictures.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Hypothetical Statement 1:
# Compare ratings of reviews with pictures and without pictures

try:
    import pandas as pd
    from scipy.stats import mannwhitneyu

    # Create a copy of the cleaned dataset
    test_data = reviews_clean.copy()

    # Convert columns into numeric format
    test_data['Rating_Numeric'] = pd.to_numeric(
        test_data['Rating'], errors='coerce'
    )
    test_data['Pictures_Numeric'] = pd.to_numeric(
        test_data['Pictures'], errors='coerce'
    )

    # Remove missing values
    test_data = test_data.dropna(
        subset=['Rating_Numeric', 'Pictures_Numeric']
    )

    # Divide reviews into two independent groups
    ratings_without_pictures = test_data.loc[
        test_data['Pictures_Numeric'] == 0, 'Rating_Numeric'
    ]

    ratings_with_pictures = test_data.loc[
        test_data['Pictures_Numeric'] > 0, 'Rating_Numeric'
    ]

    # Check whether both groups contain valid data
    if len(ratings_without_pictures) == 0 or len(ratings_with_pictures) == 0:
        raise ValueError("Insufficient data in one of the groups.")

    # Perform Mann-Whitney U Test
    statistic, p_value = mannwhitneyu(
        ratings_without_pictures,
        ratings_with_pictures,
        alternative='two-sided'
    )

    # Display the results
    print("Mann-Whitney U Test Results")
    print("--------------------------------")
    print("Reviews without pictures:", len(ratings_without_pictures))
    print("Reviews with pictures:", len(ratings_with_pictures))
    print("Average rating without pictures:",
          round(ratings_without_pictures.mean(), 2))
    print("Average rating with pictures:",
          round(ratings_with_pictures.mean(), 2))
    print("U Statistic:", round(statistic, 2))
    print("P-Value:", p_value)

    # Interpret the result using a 0.05 significance level
    if p_value < 0.05:
        print("Result: Reject the Null Hypothesis (H0).")
        print("There is a statistically significant difference between the groups.")
    else:
        print("Result: Fail to reject the Null Hypothesis (H0).")
        print("There is no statistically significant difference between the groups.")

except Exception as e:
    print("Error while performing Hypothesis Test 1:", e)
````

##### Which statistical test have you done to obtain P-Value?

**Answer Here :** I performed the Mann-Whitney U Test to obtain the P-Value and compare customer ratings between reviews with pictures and reviews without pictures.

##### Why did you choose the specific statistical test?

**Answer Here :** I chose the Mann-Whitney U Test because I wanted to compare two independent groups: reviews with pictures and reviews without pictures. This test is suitable for comparing two independent groups when the data may not follow a normal distribution. It helps determine whether the difference between the two groups is statistically significant.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Answer Here :** **Null Hypothesis (H₀):** There is no significant difference in customer rating distributions across the seven days of the week.

**Alternate Hypothesis (H₁):** At least one day of the week has a significantly different customer rating distribution.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

import pandas as pd
from scipy.stats import kruskal

# Convert Rating into numeric values
reviews_clean["Rating_Numeric"] = pd.to_numeric(
    reviews_clean["Rating"], errors="coerce"
)

# Convert Time into datetime format
reviews_clean["Review_Date"] = pd.to_datetime(
    reviews_clean["Time"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

# Extract day name
reviews_clean["Weekday"] = reviews_clean["Review_Date"].dt.day_name()

# Remove missing values
test_data = reviews_clean[["Weekday", "Rating_Numeric"]].dropna()

# Create rating groups for each weekday
days = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

rating_groups = [
    test_data[test_data["Weekday"] == day]["Rating_Numeric"]
    for day in days
]

# Perform Kruskal-Wallis H Test
statistic, p_value = kruskal(*rating_groups)

print("Total Valid Reviews:", len(test_data))
print("\nReviews and Average Ratings by Day:")

print(
    test_data.groupby("Weekday")["Rating_Numeric"]
    .agg(["count", "mean"])
    .reindex(days)
)

print("\nKruskal-Wallis H Statistic:", statistic)
print("P-value:", p_value)

# Conclusion
if p_value < 0.05:
    print("\nReject the Null Hypothesis (H0).")
    print("At least one weekday has a significantly different rating distribution.")
else:
    print("\nFail to Reject the Null Hypothesis (H0).")
    print("There is no significant difference in rating distributions across weekdays.")
````

##### Which statistical test have you done to obtain P-Value?

**Answer Here :** I have performed the Kruskal-Wallis H Test to obtain the p-value.

##### Why did you choose the specific statistical test?

**Answer Here :** I chose the Kruskal-Wallis H Test because it is used to compare the distributions of three or more independent groups. In this analysis, customer ratings are compared across seven weekdays. It does not require the data to follow a normal distribution.
