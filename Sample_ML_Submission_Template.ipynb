## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Data Wrangling Code

# 1. Remove Duplicate Rows
restaurants.drop_duplicates(inplace=True)
reviews.drop_duplicates(inplace=True)

# 2. Clean Column Names
restaurants.columns = restaurants.columns.str.strip()
reviews.columns = reviews.columns.str.strip()

# 3. Remove Extra Spaces from Text Columns
for column in restaurants.select_dtypes(include='object').columns:
    restaurants[column] = restaurants[column].str.strip()

for column in reviews.select_dtypes(include='object').columns:
    reviews[column] = reviews[column].str.strip()

# 4. Replace Empty Strings with NaN
restaurants.replace(r'^\s*$', np.nan, regex=True, inplace=True)
reviews.replace(r'^\s*$', np.nan, regex=True, inplace=True)

# 5. Handle Missing Values in Text Columns
restaurants.fillna({
    'Collections': 'Unknown',
    'Cuisines': 'Unknown',
    'Timings': 'Unknown'
}, inplace=True)

reviews.fillna({
    'Reviewer': 'Unknown',
    'Review': 'No Review Provided',
    'Metadata': 'Unknown'
}, inplace=True)

# 6. Display Cleaned Dataset Information
print("Restaurant Dataset After Data Wrangling:")
restaurants.info()

print("\nReviews Dataset After Data Wrangling:")
reviews.info()

# 7. Check Missing Values Again
print("\nRestaurant Dataset Missing Values:")
print(restaurants.isnull().sum())

print("\nReviews Dataset Missing Values:")
print(reviews.isnull().sum())

# 8. Check Duplicate Rows Again
print("\nRestaurant Dataset Duplicates:", restaurants.duplicated().sum())
print("Reviews Dataset Duplicates:", reviews.duplicated().sum())
````

#### Chart - 8

````python
# Chart - 8 visualization code
# Chart - 8: Most Frequently Used Words in Reviews

from collections import Counter
import re

# Combine all reviews into one text
all_reviews = ' '.join(reviews['Review'].dropna().astype(str))

# Convert text to lowercase
all_reviews = all_reviews.lower()

# Extract words
words = re.findall(r'\b[a-z]+\b', all_reviews)

# Remove common words (stopwords)
stop_words = [
    'the', 'is', 'and', 'a', 'to', 'of', 'in', 'it', 'was',
    'for', 'with', 'this', 'that', 'on', 'my', 'we', 'they',
    'are', 'very', 'you', 'at', 'have', 'had', 'be', 'but',
    'not', 'as', 'so', 'our', 'i', 'he', 'she', 'were',
    'from', 'their', 'me', 'been', 'food', 'restaurant'
]

filtered_words = [word for word in words if word not in stop_words]

# Find the 15 most common words
word_counts = Counter(filtered_words)
top_words = word_counts.most_common(15)

# Create DataFrame
word_df = pd.DataFrame(top_words, columns=['Word', 'Frequency'])

# Plot the chart
plt.figure(figsize=(12, 6))
sns.barplot(data=word_df, x='Frequency', y='Word')

plt.title('Top 15 Most Frequently Used Words in Reviews')
plt.xlabel('Frequency')
plt.ylabel('Words')

plt.show()
````

##### 1. Why did you pick the specific chart?

**Answer Here :** A horizontal bar chart was selected to display the most frequently used words in customer reviews. It makes it easy to compare word frequencies and identify the words that appear most often in customer feedback.

##### 2. What is/are the insight(s) found from the chart?

**Answer Here :** The chart identifies the most frequently used words in customer reviews and highlights the topics customers commonly discuss. It helps understand recurring themes in customer feedback, including food quality, service, ambience, and overall experience.

````python
# Chart - 9 visualization code
# Chart - 9: Positive vs Negative Reviews

# Convert ratings into numeric values
reviews['Rating_Numeric'] = pd.to_numeric(
    reviews['Rating'].astype(str).str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Classify reviews based on ratings
reviews['Review_Sentiment'] = reviews['Rating_Numeric'].apply(
    lambda x: 'Positive' if x >= 3.5
    else ('Negative' if x < 3.5 else 'Unknown')
)

# Count sentiment categories
sentiment_counts = reviews['Review_Sentiment'].value_counts()

# Plot the chart
plt.figure(figsize=(8, 6))
sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values)

plt.title('Positive vs Negative Reviews')
plt.xlabel('Review Sentiment')
plt.ylabel('Number of Reviews')

plt.show()
````

````python
# Chart - 10 visualization code
# Chart - 10: Top 10 Restaurants with Highest Average Ratings

# Convert ratings into numeric values
reviews['Rating_Numeric'] = pd.to_numeric(
    reviews['Rating'].astype(str).str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Calculate average rating for each restaurant
restaurant_ratings = (
    reviews.groupby('Restaurant')['Rating_Numeric']
    .mean()
    .dropna()
)

# Select top 10 restaurants by average rating
top_rated = restaurant_ratings.sort_values(
    ascending=False
).head(10)

# Plot the chart
plt.figure(figsize=(12, 6))

sns.barplot(x=top_rated.values, y=top_rated.index)

plt.title('Top 10 Restaurants with Highest Average Ratings')
plt.xlabel('Average Rating')
plt.ylabel('Restaurant Name')

plt.show()
````
